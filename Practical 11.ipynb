Represent a real-world problem using propositional logic or first-
order logic and perform logical inference to derive conclusions
using an AI reasoning approach

```python
facts = {"raning"}
print("Initial facts:", facts)
```

```python
rules = {
    "raning": "road_wet",
    "road_wet": "drive_carefully",
    "drive_carefully": "drive_speed"
}
```

```python
changed = True

while changed:
    changed = False

    for condition, conclusion in rules.items():
        if condition in facts and conclusion not in facts:
            facts.add(conclusion)
            changed = True
```

```python
print("Facts after inference:")
for fact in facts:
    print(fact)
```
