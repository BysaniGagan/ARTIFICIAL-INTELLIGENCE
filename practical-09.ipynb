Implement a simple planning problem using STRIPS / goal-stackplanning for Blocks World, logistics, or robot navigation.

```python
initial_state ={"At_A"}
goal = "At_C"
```

```python
actions = [
    {
        "name": "Move A to B",
        "preconds": {"At_A"},
        "add": {"At_B"},
        "delete": {"At_A"}
    },
    {
        "name": "Move B to C",
        "preconds": {"At_B"},
        "add": {"At_C"},
        "delete": {"At_B"}
    }
]
```

```python
def find_action(goal):
  for action in actions:
    if goal in action["add"]:
      return action
  return None
```

```python
def make_plan(goal, state, plan):
  if goal in state:
    return True

  action = find_action(goal)

  if action is None:
    return False

  for condition in action["preconds"]:
    if not make_plan(condition, state, plan):
      return False

  # Perform the action
  state.difference_update(action["delete"])
  state.update(action["add"])
  plan.append(action["name"])

  return True
```

```python
state = initial_state.copy()
plan = []

make_plan(goal, state, plan)

print("Plan:")
for step in plan:
  print(step)
```
